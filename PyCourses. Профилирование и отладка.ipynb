# 17. Профилирование и отладка

## 17.1 Отладка

***Отладка кода (debugging)*** — *это процесс поиска, локализации и исправления ошибок (багов) в программном обеспечении.* 

Если написание кода — это созидание, то отладка — это детективная работа, цель которой — понять, почему программа ведет себя не так, как ожидалось.

Отладка помогает выявить причины возникновения:
<ul>
    <li>логических ошибок: когда код работает без вылета, но выдает неверный результат.</li>
    <li>фатальных ошибок (Runtime errors): когда программа аварийно завершается до\после получения результата.</li>
</ul>

Иногда отладка может быть использована не по прямому назначению — поиск ошибок, — а для лучшего понимания принципов работы чужого кода.

Отладка в Python может быть выполнена с помощью разных инструментов:
<ul>
<li>print\logging в "проблемных" местах кода.</li>
<li>Модуль pdb.</li>
<li>Графический интерфейс среды разработки (IDE).</li>
</ul>

### 17.1.1 Примитивная отладка через print

```python
# Код с логической ошибкой и поиском проблемы через print

users = [
    {"name": "Алиса", "active": False},
    {"name": "Боб", "active": False},
    {"name": "Иван", "active": True},
    {"name": "Елена", "active": True},
]

def remove_inactive(users_list):
    print(f"Список до начала цикла: {users_list}\n")
    for user in users_list:
        print(f"Переменная в цикле: {user}\nСостояние списка до удаления элемента: {users_list}")
        if not user["active"]:
            users_list.remove(user)
        print(f"Состояние списка после удаления элемента: {users_list}\n")
    return users_list

result = remove_inactive(users)
print(result)
```

**В большинстве случаев стоит отказаться от использования отладки с помощью print в пользу более продвинутых инструментов.**

### 17.1.2 Отладка через pdb

***pdb (Python Debugger)*** — *это встроенный в стандартную библиотеку Python модуль, который представляет собой интерактивный отладчик кода для командной строки.*

pdb позволяет:
<ul>
<li>приостановить выполнение программы в любой момент (с помощью <i>точек остановки</i> breakpoint)</li>
<li>выполнить инспекцию состояния памяти в процессе отладки</li>
<li>пошагово исполнять каждую строку исходного кода</li>
<li>изменять значения переменных без выхода из режима отладки</li>
<li>анализировать состояние программы уже после её аварийного завершения</li>
</ul>


***Точка остановки (breakpoint)*** — *это преднамеренное прерывание выполнения программы в определенной строке кода, установленное программистом для проведения отладки. При достижении этой точки программа замирает, позволяя разработчику изучить текущее состояние переменных, содержимое памяти и пошагово продолжить выполнение кода.*

Для управления режимом отладки в pdb используются команды:
<ul>
<li>s(tep) - выполнить текущую строку кода и остановиться</li>
<li>c(ontinue) - продолжить выполнение до следующей <i>контрольной точки</i> (breakpoint)</li>
<li>n(ext) - продолжить выполнение до тех пор, пока не будет достигнута следующая строка в функции или не произойдет выход из функции</li>
<li>p(rint) название_переменной - показать текущее значение переменной</li>
<li>r(eturn) - продолжить выполнение до выхода из текущей функции</li>
<li>q(uit) - выйти из отладчика</li>
<li>w(here) - отобразить стек вызова до ошибки (используется при анализе работы кода при фатальной ошибке)</li>
</ul>

```python
# Код с логической ошибкой и поиском проблемы через pdb
import pdb

users = [
    {"name": "Алиса", "active": False},
    {"name": "Боб", "active": False},
    {"name": "Иван", "active": True},
    {"name": "Елена", "active": True},
]

def remove_inactive(users_list):
    for user in users_list:
        if not user["active"]:
            users_list.remove(user)
    return users_list

def main():
    result = remove_inactive(users)
    print(result)


pdb.run("main()")
```

```python
# Код с фатальной ошибкой и анализом через pdb
import pdb

users = (
    {"name": "Алиса", "active": False},
    {"name": "Боб", "active": False},
    {"name": "Иван", "active": True},
    {"name": "Елена", "active": True},
)

def remove_inactive(users_list):
    for user in users_list:
        if not user["active"]:
            users_list.remove(user)
    return users_list

def main():
    result = remove_inactive(users)
    print(result)


try:
    main()
except Exception:
    pdb.post_mortem()
```

## 17.2 Профилирование кода

***Профилирование кода (profiling)*** — *это форма анализа программы, которая используется для измерения её характеристик производительности.*

Если отладка ищет ошибки, то профилирование ищет «узкие места» (*bottlenecks*).

Профилирование позволяет измерить:
<ul>
<li>время выполнения: сколько времени тратится на работу каждой функции или отдельной строки кода.</li>
<li>использование памяти: какие части программы потребляют больше всего ОЗУ.</li>
<li>частота вызовов: сколько раз вызывается та или иная функция (часто лишние повторные вызовы тормозят систему).</li>
<li>затраты ресурсов: использование процессора (CPU) или сетевого трафика.</li>
</ul>

Важно понимать, что **профилирование ради профилирования бессмысленно**. Его стоит проводить только тогда, когда стоит задача **оптимизировать работу** кода.

### 17.2.1 Немного об оптимизации

***Оптимизация кода*** — *это процесс изменения программы для того, чтобы она работала более эффективно, при этом сохраняя свою исходную функциональность.*

Многие начинающие разработчики стремятся начать оптимизацию ещё до того, как реализовали весь функционал приложения. Это **вредное стремление**.

При упоминании словосочетания «оптимизация кода» следует вспоминать цитату известного информатика и лауреата премии Тьюринга Дональда Кнута: «**Преждевременная оптимизация — корень всех зол**».

Прежде чем переходить к какой-либо оптимизации необходимо проверить себя на основе простого чек-листа:
<ol>
<li>Было ли выполнено полное тестирование кода? Можно ли утверждать, что весь требуемый функционал работает так, как было задумано?</li>
<li>Нужно ли выполнить <i>рефакторинг</i> кода? Т.е. нет ли потребности переписать некоторые блоки кода в более читабельный и поддерживаемый вид?</li>
<li>Были ли выявлены в коде все "бутылочные горлышки" (узкие места)? Т.е. было ли выполнено профилирование?</li>
</ol>

Если код полностью протестирован, переписан в "хороший" вид и в нем выявлены все проблемные места, можно переходить к оптимизации.

### 17.2.2 Инструменты профилирования кода Python

Выполнить профилирование кода в Python можно разными инструментами:
<ul>
<li>модуль time - для замера времени выполнения (без учета "помех" в виде загруженности процессора и\или скорости работы сети)</li>
<li>модуль timeit - для замера усредненного времени выполнения (т.е. нивелируя вклад помех)</li>
<li>модуль cProfile - для получения подробных отчетов о времени выполнения кода (и каждой отдельной функции) и затраченных ресурсах</li>
</ul>

### 17.2.2.1 Примитивная фиксация времени (time)

```python
from random import randint
import time

def get_data(size: int) -> tuple[int]:
    return tuple(randint(-size, size) for _ in range(size))

def search(elements: tuple[int], element: int) -> int:
    try:
        return elements.index(element)
    except ValueError:
        return -1

def do_job(size: int) -> None:
    elements = get_data(size)
    search(elements, randint(-size, size))


t1 = time.perf_counter()
do_job(100_000)
t2 = time.perf_counter()
delta = t2 - t1

print(f"Потрачено {delta} секунд")
```

Модуль time подходит только для простой оперативной оценки производительности «здесь и сейчас», так как учитывает реальную нагрузку на компьютер. Но для высокоточных тестов, где нужно исключить влияние фоновых процессов, лучше использовать другой инструмент. С time можно снабдить каждую функцию замером времени, но это слишком трудозатратно.

### 17.2.2.2 Усредненные показатели времени (timeit)

timeit — это стандартная библиотека для точного бенчмаркинга (сравнительного анализа) производительности. Она лучше всего подходит для измерения скорости работы алгоритмов или отдельных строк кода, когда нужно получить чистый результат без влияния внешней системной среды.

Чаще всего модуль timeit используется для оценки времени выполнения небольших фрагментов кода. Для замеров используется функция **timeit**, которая принимает параметры:
<ul>
<li>stmt - код, который анализируется</li>
<li>setup - подготовительный код (его время не учитывается при оценке)</li>
<li>number - количество повторений stmt</li>
</ul>

Функция timeit всегда возвращает общее время, поэтому для получения среднего необходимо выполнить деление на количество итераций.

```python
import timeit


setup = """
from random import  randint

def get_data(size: int) -> tuple[int]:
    return tuple(randint(-size, size) for _ in range(size))

def search(elements: tuple[int], element: int) -> int:
    try:
        return elements.index(element)
    except ValueError:
        return -1

def do_job(size: int) -> None:
    elements = get_data(size)
    search(elements, randint(-size, size))
"""

stmt = "do_job(100_000)"
number = 1000

measurment = timeit.timeit(stmt, setup=setup, number=number)
print(f"Общее время: {measurment}\nВремя на 1 итерацию: {measurment / number}")
```

### 17.2.2.3 Полноценное профилирование (cProfile)

***cProfile*** — *это встроенный в Python модуль, который измеряет время выполнения каждой функции в программе и подсчитывает количество их вызовов. Он помогает найти «узкие места», предоставляя детальную статистику по всей иерархии вызовов.*

После запуска cProfile сгенерирует таблицу. Вот что значат основные столбцы:
<ul>
<li>ncalls: количество вызовов функции.</li>
<li>tottime: общее время, проведенное в самой функции (исключая время вызовов других подфункций).</li>
<li>percall: среднее время за один вызов (tottime / ncalls).</li>
<li>cumtime: кумулятивное (суммарное) время: сколько времени заняло выполнение этой функции вместе со всеми функциями, которые она вызывала внутри себя.</li>
<li>filename:lineno(function): имя файла, номер строки и название функции.</li>
</ul>

```python
from random import randint
import cProfile

def get_data(size: int) -> tuple[int]:
    return tuple(randint(-size, size) for _ in range(size))

def search(elements: tuple[int], element: int) -> int:
    try:
        return elements.index(element)
    except ValueError:
        return -1

def do_job(size: int) -> None:
    elements = get_data(size)
    search(elements, randint(-size, size))


cProfile.run("do_job(10_000)")
```
