## 1. ВВЕДЕНИЕ

### Что такое философия программирования?

Философия языка программирования — это набор принципов и убеждений, которые определяют, как язык должен использоваться и как он был спроектирован.

- Понимание философии помогает писать более качественный код
- Облегчает изучение языка
- Помогает принимать правильные решения при разработке
- Объясняет "почему" в архитектурных решениях

### Python: язык с явной философией

Python — один из немногих языков, где философия явно задокументирована и доступна прямо в интерпретаторе!

### Основные принципы Zen of Python:

#### 1. Beautiful is better than ugly

Красота важна. Код должен быть эстетически приятным.

````python
# Плохо (ugly)
def calc(x,y):return x*x+y*y

# Хорошо (beautiful)
def calculate_sum_of_squares(x, y):
    return x * x + y * y

print(calculate_sum_of_squares(3, 4))
````

#### 2. Explicit is better than implicit

Явное лучше неявного. Код должен быть понятным без догадок.

````python
# Неявно - непонятно, что происходит
# from module import *
# result = some_function()

# Явно - сразу видно, откуда функция
from math import sqrt, pi
result = sqrt(16) + pi
print(f"Результат: {result}")
````

#### 3. Simple is better than complex

Простота лучше сложности.

````python
# Сложно
def is_even_complex(n):
    return True if n % 2 == 0 else False

# Просто
def is_even(n):
    return n % 2 == 0

print(f"is_even(4): {is_even(4)}")
print(f"is_even(5): {is_even(5)}")
````

#### 4. Complex is better than complicated

Сложность лучше запутанности.

Иногда решение может быть сложным, но оно должно быть понятным, а не запутанным.

#### 5. Readability counts

Читаемость имеет значение.

Код читается гораздо чаще, чем пишется. Поэтому читаемость — приоритет #1.

````python
# Плохо читается
x=[i*2 for i in range(10) if i%2==0]
print(f"Плохо: {x}")

# Хорошо читается
even_numbers = [i * 2 for i in range(10) if i % 2 == 0]
print(f"Хорошо: {even_numbers}")
````

#### 6. There should be one—and preferably only one—obvious way to do it

Должен быть один — и желательно только один — очевидный способ сделать это.

В отличие от Perl ("There's more than one way to do it"), Python предпочитает единственный правильный путь.

#### 7. Errors should never pass silently

Ошибки никогда не должны проходить молча.

````python
# Плохо - ошибка игнорируется
try:
    result = 10 / 0
except:
    pass  # Что-то пошло не так, но мы не знаем что

# Хорошо - ошибка обрабатывается явно
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Ошибка деления на ноль: {e}")
````

#### 8. In the face of ambiguity, refuse the temptation to guess

При неоднозначности откажитесь от соблазна угадывать.

Python не будет угадывать ваши намерения. Если что-то неоднозначно, вы получите ошибку.

#### 9. If the implementation is hard to explain, it's a bad idea

Если реализацию сложно объяснить, это плохая идея.

Если вы не можете объяснить свой код, возможно, стоит его упростить.

#### 10. If the implementation is easy to explain, it may be a good idea

Если реализацию легко объяснить, это может быть хорошая идея.

### 3.2. Duck Typing ("Если это ходит как утка и крякает как утка...")

Не важен тип объекта, важны его методы и атрибуты.

````python
class Duck:
    def quack(self):
        return "Quack!"

class Person:
    def quack(self):
        return "I'm quacking like a duck!"

def make_it_quack(thing):
    # Нас не интересует тип, только наличие метода quack()
    print(thing.quack())

make_it_quack(Duck())    # Работает
make_it_quack(Person())  # Тоже работает!
````

### 3.3. EAFP: Easier to Ask for Forgiveness than Permission

Проще просить прощения, чем разрешения.

````python
my_dict = {"name": "Python", "version": "3.11"}

# EAFP стиль (Python way)
try:
    value = my_dict['key']
except KeyError:
    value = 'default'
    print(f"EAFP: {value}")

# LBYL стиль (Look Before You Leap)
if 'key' in my_dict:
    value = my_dict['key']
else:
    value = 'default'
    print(f"LBYL: {value}")
````

### 3.4. Интуитивность и предсказуемость

Python стремится быть интуитивным.

````python
# Интуитивные операции
numbers = [1, 2, 3, 4, 5]
squares = [x ** 2 for x in numbers]  # Список квадратов
evens = [x for x in numbers if x % 2 == 0]  # Чётные числа

print(f"Числа: {numbers}")
print(f"Квадраты: {squares}")
print(f"Чётные: {evens}")

# Читается почти как английский язык!
````

### 3.5. Минимум синтаксического сахара

## 4. ПРАКТИЧЕСКИЕ ПРИМЕРЫ

### Пример 1: Читаемость кода

**Задача:** Найти все чётные числа и возвести их в квадрат.

````python
# Плохо (не следует философии Python)
def f(l):
    r=[]
    for i in l:
        if i%2==0:
            r.append(i*i)
    return r

print(f"Плохо: {f([1, 2, 3, 4, 5, 6])}")

# Хорошо (следует философии Python)
def get_squares_of_even_numbers(numbers):
    return [number ** 2 for number in numbers if number % 2 == 0]

print(f"Хорошо: {get_squares_of_even_numbers([1, 2, 3, 4, 5, 6])}")
````

### Пример 2: Явность лучше неявности

````python
# Неявно - непонятно, что происходит
# from math import *
# result = sqrt(16) + pi

# Явно - сразу видно источник
from math import sqrt, pi
result = sqrt(16) + pi
print(f"Результат: {result}")
````

### Пример 3: Простота

````python
# Сложно
def get_max_complex(numbers):
    if len(numbers) == 0:
        return None
    max_value = numbers[0]
    for num in numbers:
        if num > max_value:
            max_value = num
    return max_value

# Просто
def get_max(numbers):
    return max(numbers) if numbers else None

test_numbers = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"Сложно: {get_max_complex(test_numbers)}")
print(f"Просто: {get_max(test_numbers)}")
````

### Пример 4: Обработка ошибок

````python
# Плохо - ошибка игнорируется
def divide_bad(a, b):
    try:
        return a / b
    except:
        return None

# Хорошо - ошибка обрабатывается явно
def divide_good(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        raise ValueError("Cannot divide by zero")

print(f"divide_bad(10, 0): {divide_bad(10, 0)}")

try:
    divide_good(10, 0)
except ValueError as e:
    print(f"divide_good(10, 0): {e}")
````
