### Принцип работы веб-приложений

Разберем полный флоу на примере: пользователь вводит в браузере `tusur.ru`.

1. **Ввод URL в адресную строку**
   - Браузер понимает, что нужно открыть сайт по доменному имени `tusur.ru`.

2. **Поиск IP-адреса через DNS**
   - Браузер/ОС проверяет кэш DNS.
   - Если записи нет, отправляется запрос к DNS-серверу.
   - DNS возвращает IP-адрес сайта (например, `x.x.x.x`).

3. **Установка соединения с сервером**
   - По найденному IP браузер открывает соединение (обычно HTTPS: TCP + TLS).

4. **Отправка HTTP-запроса**
   - Для открытия страницы обычно отправляется `GET /`.
   - Сервер обрабатывает запрос и возвращает ответ html-код.

5. **Получение HTTP-ответа**
   - Сервер возвращает статус (например, `200`) и тело ответа (часто HTML).

6. **Отрисовка страницы в браузере**
   - Браузер парсит HTML.
   - Находит дополнительные ресурсы (`CSS`, `JS`, изображения, шрифты) и загружает их отдельными запросами.
   - Строит DOM/CSSOM и отрисовывает страницу.

7. **Выполнение JavaScript и динамика**
   - JS может реагировать на действия пользователя (клик, ввод), менять DOM и делать новые запросы к серверу (например, `fetch('/api/news/')`), получая JSON и обновляя часть страницы без полной перезагрузки.

![Полный поток работы браузера, сервера и базы данных](img/browser_full_flow.png)

Пример динамики на JavaScript (без перезагрузки страницы):

```html
<button id="load-news">Показать новости</button>
<ul id="news"></ul>

<script>
document.getElementById("load-news").addEventListener("click", async () => {
  const response = await fetch("/api/news/");
  const items = await response.json();
  document.getElementById("news").innerHTML = items
    .map((item) => `<li>${item.title}</li>`)
    .join("");
});
</script>
```

Базовые HTTP-методы:
- `GET` - получить данные (страница, JSON, файл);
- `POST` - отправить данные на сервер (например, форму).

Примеры HTTP-статусов:
- `200 OK` - запрос выполнен успешно;
- `404 Not Found` - страница не найдена;
- `500 Internal Server Error` - ошибка на стороне сервера.

### Что такое фреймворк и зачем он нужен

**Фреймворк** - это каркас для разработки приложений: он задает структуру проекта и уже содержит готовые инструменты для типовых задач.

Зачем использовать фреймворк в web-разработке:
- не писать с нуля повторяющиеся вещи (маршрутизация, работа с формами, авторизация);
- ускорить старт проекта и разработку новых функций;
- уменьшить количество ошибок за счет проверенных решений;
- упростить командную разработку благодаря единому стилю структуры проекта.

### Почему именно Django

Django - это высокоуровневый Python web-фреймворк для быстрой и безопасной разработки.

Исторически Django создавался для разработки новостных сайтов с частыми изменениями контента и жесткими сроками публикации. Поэтому в его основе лежит подход:
- **быстро сделать рабочий продукт**;
- **не дублировать код (DRY)**;
- **изначально учитывать безопасность**.

### Какие проблемы Django помогает решать

Django закрывает типичные backend-задачи «из коробки»:
- маршрутизация URL и обработка HTTP-запросов;
- работа с базой данных через ORM (без ручного SQL в большинстве случаев);
- админ-панель для управления данными;
- формы и валидация пользовательских данных;
- аутентификация и права доступа;
- базовая защита от распространенных уязвимостей (CSRF, XSS, SQL Injection).

Итог: Django позволяет сосредоточиться на бизнес-логике проекта, а не на инфраструктурной «обвязке».

**Фреймворк Django** – это интегрированная система (или среда) для разработки веб-приложений на языке Python. Один из основных принципов фреймворка – DRY (англ. Don't repeat yourself).

### URL -> View -> Template (ядро Django)

В классическом Django-приложении запрос проходит через три ключевых этапа:

1. **URL** (`urls.py`) - определяет, какой обработчик вызвать для конкретного пути.
2. **View** (`views.py`) - содержит бизнес-логику: получить данные, проверить условия, подготовить контекст.
3. **Template** (`templates/*.html`) - отвечает за отображение данных пользователю.

Упрощенный поток:
- пользователь открывает `/students/`;
- Django находит маршрут в `urls.py`;
- вызывается функция/класс во `views.py`;
- view получает данные из моделей и передает их в шаблон;
- шаблон рендерится в HTML и возвращается в браузер.

Это разделение помогает писать поддерживаемый код: маршруты отдельно, логика отдельно, интерфейс отдельно.

Архитектура Django похожа на «Модель-Представление-Шаблон» (MTV), которую часто сравнивают с MVC.

Контроллер классической модели MVC примерно соответствует уровню, который в Django называется Представление (**views.py**), а презентационная логика представления реализуется уровнем Шаблонов (**templates**).

### Виртуальное окружение (venv)

**Зачем нужно:**
- виртуальное окружение - это отдельная «папка с Python-пакетами» для одного проекта;
- пакеты из одного проекта не конфликтуют с пакетами другого проекта;
- проще воспроизводить проект у преподавателя/студента/на другом компьютере;
- безопаснее обновлять библиотеки, не ломая глобальный Python.

**Как пользоваться (базовый сценарий):**
1. Один раз создать окружение в папке проекта.
2. Перед работой активировать окружение.
3. Устанавливать пакеты уже внутри окружения.
4. После установки сохранить зависимости в `requirements.txt`.
5. После завершения работы при необходимости выйти из окружения (`deactivate`).

**Основные команды (macOS/Linux):**
```bash
# создание окружения
python -m venv .venv

# активация окружения
source .venv/bin/activate

# обновить pip и установить Django
pip install --upgrade pip
pip install Django

# сохранить зависимости проекта
pip freeze > requirements.txt

# установить зависимости из файла (на другом ПК)
pip install -r requirements.txt

# выход из окружения
deactivate
```

**Основные команды (Windows PowerShell):**
```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install --upgrade pip
python -m pip install "Django==6.0.*"
python -m pip freeze > requirements.txt
python -m pip install -r requirements.txt
deactivate
```

**Основные команды Django**:
- создать новый проект: **django-admin startproject example**
- перейти в папку проекта: **cd example**
- применить миграции: **python manage.py migrate**
- запустить локальный web-сервер: **python manage.py runserver**
- создать приложение: **python manage.py startapp university**
- создать суперпользователя для admin: **python manage.py createsuperuser**

### Модели и БД как основа backend

Модель в Django - это Python-класс, который описывает структуру таблицы базы данных.

Почему это основа backend:
- данные приложения хранятся в БД;
- правила целостности (типы полей, связи, ограничения) задаются в моделях;
- большая часть бизнес-логики опирается на чтение и изменение данных.

Базовые операции backend через ORM:
- **Create** - создать запись;
- **Read** - получить данные;
- **Update** - изменить данные;
- **Delete** - удалить данные.

Изменения в моделях применяются через миграции:
- `python manage.py makemigrations`
- `python manage.py migrate`

Ниже приведен базовый пример модели и основных ORM-операций.

Django предоставляет разработчикам возможность использовать **ORM**.

**ORM (object-relational mapping, объектно-реляционное отображение)** - технология программирования, которая связывает базы данных с концепциями объектно-ориентированных языков программирования, создавая «виртуальную объектную базу данных».

К преимуществам ORM можно отнести:
 - *переносимость*,
 - *безопасность*,
 - *независимость от СУБД*,
 - *простота применения*.

Все модели (таблицы в базе данных) представляют собой класс (обычно в `models.py`), наследующий `django.db.models.Model`.

> Примечание: в современных версиях Django первичный ключ по умолчанию - `BigAutoField`, поэтому точный SQL-тип `id` зависит от СУБД и версии Django.

Пример создания модели Student (приближенный SQL-эквивалент):
```sql
CREATE TABLE app_student (
    id bigint NOT NULL PRIMARY KEY,
    name varchar(30) NOT NULL,
    email varchar(254) NULL,
    group_number integer NOT NULL
);
```

````python
from django.db import models


class Student(models.Model):
    name = models.CharField(max_length=30)
    email = models.EmailField(blank=True)
    group_number = models.IntegerField()
````

#### Основные команды и методы при работе с ORM

**Создание миграций**: `python manage.py makemigrations`

**Применение миграций**: `python manage.py migrate`

**Запуск Django shell**: `python manage.py shell`

**Создание объекта** (записи в таблице):
```python
Student.objects.create(name="Ivanov I.I.", email="ivanov@ivanovy.com", group_number=128)
```

**Удаление объектов**:
```python
Student.objects.filter(pk=1).delete()
```

**Обновление объектов**:
```python
Student.objects.filter(pk=1).update(name="Ivanov F.F.")
```

**Получение всех объектов** (записей из таблицы):
```python
Student.objects.all()
```

**Получение определенных объектов** (фильтрация):
```python
Student.objects.filter(group_number=128)
```

Документация (всегда актуальная ветка): https://docs.djangoproject.com/en/stable/topics/db/

1. Создать репозиторий на GitHub.
2. Создать и активировать виртуальное окружение (`venv`).
3. Установить Django (рекомендуется фиксировать версию: `Django==6.0.*` или `Django==5.2.*`) и сохранить зависимости в `requirements.txt`.
4. Создать Django-проект и настроить систему контроля версий для использования репозитория, созданного ранее.

8. Разместить созданный проект в репозитории. Рекомендуется вносить изменения регулярно (не менее 3-4 коммитов).
